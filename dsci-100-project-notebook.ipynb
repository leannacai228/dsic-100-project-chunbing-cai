#### Task 3.1: Tidy the data (5 points)

Produce a tidy data set that could be used to answer your research question using the `tidyverse`. Please ensure you have at least one quantitative (numeric) and one categorical variable:
- Create new variables (if needed)
- Transform the data into a tidy form (if needed)
- Remove irrelevant columns (if needed)
- Comment your code throughout
- Show the first 6 rows of the tidied data.

</div>

```r
march_madness_tidy <- march_madness |>
    select(seed, reg_wins_pct, tourney_wins) |>
    drop_na() |>
    filter(seed > 0) |>
    mutate(win_category = as.factor(if_else
                                    (tourney_wins >= 1, "at_least_one_win", "no_win"))) |>
    select(seed, reg_wins_pct, win_category)

head(march_madness_tidy, 6)
    
```

#### Task 4.1: Calculate summary statistics of numeric variables (2 points)

```r
# Calculate summary statistics for all numeric columns
numeric_summary <- march_madness_tidy |>
  summarise(across(where(is.numeric),list(
        mean = ~ mean(.x, na.rm = TRUE),
        minimum = ~ min(.x, na.rm = TRUE),
        maximum = ~ max(.x, na.rm = TRUE)),
      .names = "{.col}_{.fn}"))

# Display the summary table
numeric_summary
```

```r
#distribution
march_madness_tidy_distribution <- ggplot(march_madness_tidy, aes(x = seed)) +
  geom_histogram(binwidth = 1,boundary = 0.5,fill = "blue")+
  scale_x_continuous(breaks = seq(
      min(march_madness_tidy$seed),
      max(march_madness_tidy$seed),
      by = 1)) +
  labs(title = "Distribution of Team Seeds",x = "Team Seed",y = "Number of Team-Year Observations") 
march_madness_tidy_distribution
```

```r
#proportion
category_proportions <- march_madness_tidy |>
  count(win_category) |>
  mutate(proportion = n / sum(n))

march_madness_tidy_proportion <- ggplot(category_proportions,
       aes(x = win_category, y = proportion)) +
  geom_col(fill = "pink") +
  labs(title = "Proportion of Tournament Outcome Categories",x = "Tournament Outcome", y = "Percentage of Team-Year Observations") 
march_madness_tidy_proportion
```

Visualization 1: This plot compares the distribution of team seeds between teams that win at least one tournament game and teams that win none. Lower seed numbers represent higher-ranked teams, and overlap between the groups indicates that seed may not perfectly distinguish the two categories.

```r
seed_relationship_graph <- march_madness_tidy |>
  ggplot(aes(x = win_category, y = seed)) +
  geom_boxplot(fill = "lightblue") +
  labs(title = "Team Seed by Tournament Outcome",
    x = "Tournament Outcome",
    y = "Team Seed (Lower Number = Higher Rank)")

seed_relationship_graph
```

Before modeling, I will verify that regional win percentage is available before the tournament and does not include tournament results. 

This method is appropriate because the outcome has two categories and both predictors are numeric. k-NN classifies a team based on the outcomes of teams with similar predictor values. 

However, it is sensitive to predictor scales and the choice of the number of neighbors, and these two predictors may not capture all factors affecting tournament outcomes. Repeated appearances by the same teams and changes across years may also limit generalization. 

After cleaning the data, I will randomly split it into 75% training data and 25% testing data, before standardizing the predictors. I will use five-fold cross-validation on the training set to choose K.

During cross-validation, each fold will be used for validation once, while the remaining four folds are used for training. Both predictors will be standardized within each training fold because k-NN relies on distances. I will compare several K values and select the one with the highest average validation accuracy.

Finally, I will fit the selected model using the entire training set and evaluate it on the untouched testing set. My main performance metric will be accuracy, the proportion of correctly classified teams. This is appropriate for a binary classification task when the classes are reasonably balanced.
